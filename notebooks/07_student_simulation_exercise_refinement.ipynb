# Student Simulation as a Refinement Loop

> First-shot generation → simulated student feedback → revised generation → human evaluation

My main idea is not to use student simulation as a pre-test or as a measurement of real learning. I want to use it as a **feedback mechanism for improving the first generation**.

The motivation comes from our own generation results: prompting can produce plausible text, but it does not reliably make the model think like a student or create genuinely open-ended PF exercises.

## 1. Problems I observed in the first generation

| Evidence from our work | Problem | Why a second generation is needed |
|---|---|---|
| Items 8, 11, 13, and 19 | False rounding conflict, inconsistent scale factors, impossible bucket constraints, and contradictory mixture totals. | The model needs feedback that distinguishes productive difficulty from invalid mathematics. |
| First Qwen pool | 21 records contained only four unique visible exercises. | The first-shot prompt did not create enough variation. |
| DeepSeek expert review | Multiple Representations and Solutions scored 1 in every condition. | The exercises remained mostly prose-based and procedural. |
| DeepSeek expert review | Open-Endedness was weak, especially in V2. | “Design” and “explain” did not produce genuinely multiple pathways. |
| Our MCQ generation experience | Adding more instructions, reasoning requests, or a student-like persona did not guarantee authentic student errors or sufficient diversity. | Prompting alone is not enough; the generator needs concrete response-based feedback. |

These findings show me that the problem is not only whether the model can generate a valid text. The problem is whether the generated exercise creates the kind of diverse and discussable student work required by Productive Failure.

## 2. Why the current evaluation is not enough

My current checks and human rubric evaluate the exercise as a text artifact. They can identify missing sections, poor wording, weak open-endedness, or mathematical problems.


## 2. Evidence for using a feedback loop

| Source | Evidence relevant to my design |
|---|---|
| **Our MCQ generation experiments** | Prompt variants such as pedagogical framing, explicit reasoning, self-critique, and a student-like persona changed the instructions, but did not guarantee authentic errors or diverse outputs. This suggests that prompting alone has a limited ceiling. |
| **Our PF review** | The fixed DeepSeek pool received `MR = 1.0` in all conditions and weak Open-Endedness scores. This shows that the first-shot generator remained inside a narrow procedural format. |
| **Kapur & Bielaczyc (2012)** | PF depends on students generating different representations and solution methods before instruction. The exercise should therefore be refined according to the response space it creates. |
| **Puech et al. (2025)** | In their simulated-student study, the controlled PF tutor produced more RSMs than the no-intent baseline on the Consistency problem: 2.6 versus 1.0. This supports using simulated interactions to improve the generation of diverse attempts, although their goal was tutor evaluation. |
| **Lu & Wang (2024)** | Their profile-based simulated students showed a correlation of about $r=0.72$ with real student response patterns on MCQs. This supports using explicit knowledge states, while also showing that simulation requires calibration. |
| **Rahimi et al. (2025)** | Their human evaluation showed that strong PF exercises score particularly well on Multiple Representations and Open-Endedness. These are exactly the dimensions that are weak in our first pool. |

Together, these findings motivate a feedback loop rather than another one-shot prompt.

## 3. My proposed role for student simulation

I do not want to claim that the model truly becomes a real student. Instead, I use simulated students to generate **alternative response patterns** that the exercise might invite.

The simulated responses give the generator something concrete to react to. They can reveal that:

- all profiles use the same direct method;
- the intended misconception never appears;
- the task does not support multiple representations;
- failure is caused by confusing or contradictory data;
- the task is too restricted to create productive exploration.

The important point is that the simulator is not the final judge. It creates feedback such as:

> “Most attempts use the same direct ratio calculation. Add a meaningful comparison or another representation while preserving the learning objective.”

The generator then uses this feedback to produce a less rigid second version. Human evaluation remains the final quality check.

## 4. What student simulation can help me solve

| Problem found in the first shot | Feedback from simulated attempts | Possible second-shot revision |
|---|---|---|
| Low Multiple Representations and Solutions | Most profiles use the same direct method. | Require comparison between methods or invite a table, diagram, estimate, or explanation. |
| Low Open-Endedness | The task has one obvious answer and one main path. | Add meaningful choices, constraints, or multiple defensible approaches. |
| Missing misconception activation | The profile has a misconception, but the response does not expose it. | Change the task so the misconception becomes relevant and visible. |
| Unproductive failure | Responses are incoherent, contradictory, or unrelated. | Recheck wording and mathematical validity before refining the pedagogy. |
| No profile effect | All learner profiles produce almost the same response. | Improve profile grounding or treat the simulation as unreliable. |
| Low diversity between generations | Different generation calls produce nearly identical tasks or methods. | Use response feedback to request genuinely different representations and pathways. |

Student simulation therefore helps the generator **escape the rigid first-shot format**. It does not replace duplicate checks, mathematical validation, or final human evaluation.

## 5. Refined generation and refinement logic

The revised methodology is not:

> Generate → pre-test → accept or reject.

It is:

> Generate → simulate alternative student responses → summarize feedback → regenerate with the feedback.

```mermaid
flowchart LR
    A[First-shot PF generation] --> B[Simulate diverse learner responses]
    B --> C[Identify rigid methods, missing RSMs, and unproductive failure]
    C --> D[Give structured feedback to the generator]
    D --> E[Second-shot refinement]
    E --> B
    E --> F[Final human evaluation]
```

The process has two safeguards:

1. **Structural and mathematical checks** prevent invalid exercises from entering the loop.
2. **Human evaluation** checks whether the refined exercise actually improves PF quality.

My central claim is therefore:

> Student simulation is a feedback-driven refinement method. It gives the generator concrete alternative attempts and weaknesses to respond to, helping it move beyond the rigid, one-path output produced by first-shot prompting.

The goal is not to optimize a simulated student score. The goal is to increase meaningful RSM diversity and Open-Endedness while preserving mathematical validity and the intended learning objective.
